## 7. Category-wise Sales Analysis

#### Total sales by category
The total sales for each category are calculated using
`groupby()` and the `sum()` aggregation function.

```python
category_sales = df.groupby("Category")["Total_Sales"].sum()
category_sales
```

```python
# Average sales by category
category_avg_sales = df.groupby("Category")["Total_Sales"].mean()
category_avg_sales
```

```python
# Total quantity sold by category
category_quantity = df.groupby("Category")["Quantity"].sum()
category_quantity
```

```python
# Number of orders by category
category_orders = df.groupby("Category")["Order_ID"].count()
category_orders
```

```python
category_analysis = df.groupby("Category").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
)

category_analysis
```

```python
# Highest-performing category
category_analysis.sort_values(by="Total_Sales", ascending=False).head(1)
```

## 8. City-wise Sales Analysis

```python
# Total quantity sold by city
city_quantity = df.groupby("City")["Quantity"].sum()
city_quantity
```

```python
# Number of orders by city
city_orders = df.groupby("City")["Order_ID"].count()
city_orders
```

```python
city_analysis = df.groupby("City").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
)

city_analysis
```

```python
# Find the highest-performing city 
city_analysis.sort_values(by="Total_Sales", ascending=False).head(1)
```

```python
 # Lowest-performing city
city_analysis.sort_values(by="Total_Sales", ascending=True).head(1)
```

## 12. 📝 Final Observations and Insights

### Key Findings

- The dataset contains 100 orders with total sales of ₹447,262.95. The average sales value per order is approximately ₹4,472.63, and a total of 294 units were sold.

- The highest individual order has a sales value of ₹13,995, while the lowest individual order has a sales value of ₹449.10.

- Among all categories, Sports is the top-performing category with total sales of ₹136,026.45. Home & Kitchen is the second-highest category with sales of ₹113,329.90.

- Sports also has the highest total quantity sold, with 79 units across 25 orders. This indicates that Sports products have a strong contribution to both sales and order volume.

- Hyderabad recorded the highest total sales among the cities, with ₹59,076.45. Lucknow also showed a relatively high average sales value of ₹7,699.11 per order.

- Coffee Maker is the highest-selling product by total sales, generating ₹51,610.25 from 7 orders. Jacket is another strong-performing product with total sales of ₹47,106.15.

- Some products have a high average sales value even with fewer orders. For example, Jacket has an average sales value of ₹9,421.23, while Mechanical Keyboard has an average of ₹8,829.80.

- Credit Card is the payment method with the highest total sales at ₹106,540.25. UPI has the highest number of orders with 26 orders, showing that it is the most frequently used payment method in the dataset.

- Net Banking has the highest average sales value among the payment methods, at approximately ₹5,668.90 per order.

- The filtering and sorting operations helped identify high-value orders, top-performing products, categories, cities, and payment methods.

### Conclusion

Overall, the analysis shows that sales performance varies across categories, cities, products, and payment methods. Sports and Home & Kitchen are the strongest categories, while Coffee Maker and Jacket are among the top-performing products. Hyderabad recorded the highest city-wise sales, and Credit Card generated the highest total sales among payment methods. Pandas functions such as filtering, sorting, `groupby()`, and aggregation were useful for finding these patterns and understanding the e-commerce sales data.
